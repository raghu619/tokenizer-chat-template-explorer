```python
MODELS = {
    "Llama-3.1":     "meta-llama/Meta-Llama-3.1-8B-Instruct",
    "Phi-4-mini":    "microsoft/Phi-4-mini-instruct",
    "DeepSeek-V3.1": "deepseek-ai/DeepSeek-V3.1",
    "Qwen-Coder":    "Qwen/Qwen2.5-Coder-7B-Instruct",
}
```

```python
from transformers import AutoTokenizer

tokenizers = {}

for name, repo in MODELS.items():
    print(f"Loading {name} ...")
    tokenizers[name] = AutoTokenizer.from_pretrained(repo, trust_remote_code=True)

print("\nAll tokenizers loaded.")
```

```python
text = "I am curiously excited to show Hugging Face Tokenizers in action to my LLM engineers"

print(f"Sentence: {text!r}")
print(f"Characters: {len(text)}, Words: {len(text.split())}\n")

for name, tok in tokenizers.items():
    ids = tok.encode(text)
    print(f"{name:<14} -> {len(ids):>3} tokens")
```

```python
text = "I am curiously excited to show Hugging Face Tokenizers in action to my LLM engineers"

for name, tok in tokenizers.items():
    ids = tok.encode(text)
    pieces = tok.batch_decode(ids)
    print(f"=== {name} ({len(ids)} tokens) ===")
    print(pieces)
    print()
```

```python
import pandas as pd

text = "I am curiously excited to show Hugging Face Tokenizers in action to my LLM engineers"
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user",   "content": "Tell a light-hearted joke for a room of Data Scientists"},
]

rows = []
for name, tok in tokenizers.items():
    raw_ids = tok.encode(text)

    chat_string = tok.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)


    chat_ids = tok.encode(chat_string)

    rows.append({
        "Model": name,
        "Vocab size": tok.vocab_size,
        "Tokens (raw sentence)": len(raw_ids),
        "Tokens (chat formatted)": len(chat_ids),
    })

df = pd.DataFrame(rows).set_index("Model")
df
```

```python
code = '''
def hello_world(person):
    print("Hello", person)
'''

qwen = tokenizers["Qwen-Coder"]
ids = qwen.encode(code)

print(f"{len(ids)} tokens\n")
for tid in ids:
    print(f"{tid:>6} = {qwen.decode([tid])!r}")
```
